```python
def vaccination_status(x):
    if pd.isna(x):
        return "No Data"
    elif x < 25:
        return "Low"
    elif x < 50:
        return "Medium"
    elif x < 75:
        return "High"
    else:
        return "Very High"

final_country["vaccination_status"] = (
    final_country["people_fully_vaccinated_per_hundred"]
    .apply(vaccination_status)
)
```
